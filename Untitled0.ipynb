```python
import math

def is_prime(n):
    # Check for the smallest numbers
    if n <= 1:
        return False
    if n <= 3:
        return True
    # Eliminate multiples of 2 and 3
    if n % 2 == 0 or n % 3 == 0:
        return False
    # Check for factors from 5 to the square root of n
    i = 5
    while i * i <= n:
        if n % i == 0 or n % (i + 2) == 0:
            return False
        i += 6
    return True
```
